```python
import os
import pandas as pd

def merge_product_data(base_dir, output_file="merged_product_data.xlsx"):
    all_dataframes = []

    # Walk through all subfolders
    for root, _, files in os.walk(base_dir):
        for file in files:
            if file == "product_data.xlsx":
                file_path = os.path.join(root, file)
                try:
                    df = pd.read_excel(file_path)
                    df['Source File'] = file_path  # Optional: helps trace back origin
                    all_dataframes.append(df)
                    print(f"Loaded: {file_path}")
                except Exception as e:
                    print(f"Failed to load {file_path}: {e}")

    if all_dataframes:
        merged_df = pd.concat(all_dataframes, ignore_index=True)
        merged_df.to_excel(output_file, index=False)
        print(f"✅ Merged file saved to: {output_file}")
    else:
        print("⚠️ No product_data.xlsx files found.")

# Example usage:
if __name__ == "__main__":
    merge_product_data(base_dir="./archive")
```
